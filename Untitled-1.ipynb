```python
import openvino_genai as ov_genai
import sys
import time

model_dir = "ov_model/"
device = "CPU"

print(f"Loading model from {model_dir}\n")
pipe = ov_genai.LLMPipeline(str(model_dir), device)
pipe.get_tokenizer().set_chat_template(pipe.get_tokenizer().chat_template)

generation_config = ov_genai.GenerationConfig()
generation_config.max_new_tokens = 2048

# Variables for tracking tokens and timing
token_count = 0
start_time = None

def streamer(subword):
    global token_count, start_time
    
    # Initialize timing on first token
    if start_time is None:
        start_time = time.time()
    
    token_count += 1
    print(subword, end="", flush=True)
    sys.stdout.flush()
    
    # Return flag corresponds whether generation should be stopped.
    # False means continue generation.
    return False

def calculate_tokens_per_second():
    """Calculate and return tokens per second"""
    global token_count, start_time
    
    if start_time is None or token_count == 0:
        return 0.0
    
    elapsed_time = time.time() - start_time
    if elapsed_time == 0:
        return 0.0
    
    return token_count / elapsed_time

def reset_token_tracking():
    """Reset token tracking variables"""
    global token_count, start_time
    token_count = 0
    start_time = None

# Example usage
input_prompt = "how many r are there in the word 'strawberry'?"
print(f"Input text: {input_prompt}")

# Reset tracking before generation
reset_token_tracking()

# Generate response
result = pipe.generate(input_prompt, generation_config, streamer)

# Calculate and display tokens per second
tps = calculate_tokens_per_second()
print(f"\n\nGeneration completed:")
print(f"Total tokens: {token_count}")
print(f"Tokens per second: {tps:.2f}")

# Reset for next generation
reset_token_tracking()
```
